```python


a = np.array([-16, 25, -36, 49])

absolute = np.abs(a)
print("Absolute values:", absolute)

square_root = np.sqrt(absolute)
print("Square roots:", square_root)
```

```python


a = np.arange(7, 13).reshape(3, 2)

print("Original matrix:")
print(a)
print("Original shape:", a.shape)

transpose = a.T

print("Transposed matrix:")
print(transpose)
print("Transposed shape:", transpose.shape)
```

```python


temperatures = np.array([18, 22, 27, 31, 24, 19])

temp_range = np.max(temperatures) - np.min(temperatures)
median = np.median(temperatures)

print("Range:", temp_range)
print("Median:", median)
```
