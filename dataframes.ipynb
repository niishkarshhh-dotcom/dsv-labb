```python
movie_data = {
    'Movie': ['3 Idiots', 'Dangal', 'PK'],
    'Year': [2009, 2016, 2014],
    'Rating': [8.4, 8.3, 8.1]
}

movie_df = pd.DataFrame(movie_data)
print(movie_df)
```

```python
new_movie_df = movie_df[['Movie', 'Rating']].copy()

new_movie_df['Director'] = [
    'Rajkumar Hirani',
    'Nitesh Tiwari',
    'Rajkumar Hirani'
]

print(new_movie_df)
```

```python
city_series = pd.Series(
    ['Delhi', 'Mumbai'],
    index=[0, 3]
)

print(city_series)
```

```python
branch_df = pd.DataFrame(index=range(4))

branch_df['Branch'] = city_series

print(branch_df)
```
